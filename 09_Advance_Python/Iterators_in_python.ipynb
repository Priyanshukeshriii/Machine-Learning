## Iterators in Python


```python
lst = [19,9,4,29]
```

```python
it = iter(lst)
```

```python
print(next(it))
print(next(it))
print(next(it))
print(next(it))
```

### How a for loop uses iterators internally

```python
for num in lst:
    print(num)
```

```python
# The above code is similary to 
it = iter(lst)

while True:
    try:
        num = next(it)
        print(num)
    except StopIteration:
        break
```

### Memory Advantage of iterators
The biggest reason iterator exist


```python
# without iterator
num = list(range(100000))
# this stores all number in memory
```

```python
# with iterator
num = iter(range(100000))

# 1. Numbers are produced one at a time
# 2. Only the current value is kept in memory
# 3. This saves a huge amount of RAM
```
