```python
# Polymorphism in Python
# * Polymorphism is used to perform a single action in different ways.
# * It allows us to override methods in a child class, providing a different implementation while maintaining the same method name.
```

```python
# base class 


class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def make_sound(self):
        print("The animal makes a sound")

# * derived class 1

class Dog(Animal):
    def __init__(self, name, species, owner, breed):
        super().__init__(name, species)
        self.owner = owner
        self.breed = breed
    
    def make_sound(self):
        print("Woof! Woof!")
    
    def play(self):
        print(f"{self.name} is playing fetch.")

# *derived class 2

class Cat(Animal):
    def __init__(self, name, species, owner, color):
        super().__init__(name, species)
        self.owner = owner
        self.color = color
    
    def make_sound(self):
        print("Meow! Meow!")
    
    def play(self):
        print(f"{self.name} is playing with a ball of yarn.")
```

### Polymorphism with Abstract base classes

Abstract Base Classes (ABCs) are used to define common method for a group of related object. They can enforce that derived classes implement particular methods , promoting consistency across different implementations

```python
from abc import ABC , abstractmethod
# define an abstract class
class vechile(ABC):
    @abstractmethod
    def drive(self):
        pass   


#  derived class 1
class Car(vechile):
    def __init__(self, windows , doors , enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
```
