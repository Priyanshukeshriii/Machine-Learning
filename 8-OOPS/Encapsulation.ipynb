Encapsulation in OOPS


```python
class Person:
    def __init__(self, name, age):
        self.name = name # ! this is public attribute
        self.age = age # !this is public attribute
```

```python
person1 = Person("priyanshu" , 22)
```

```python
person1.age = 23
print(person1.age)
```

```python
class Emoployee:
    def __init__(self, name, age):
        self.__name = name # ! this is private attribute
        self.__age = age # !this is private attribute
    
    def get_name(self):
        return self.__name
    def get_age(self):
        return self.__age
```

```python
class Employee:
    def __init__(self, name, age):
        self._name = name # ! this is protected attribute
        self._age = age # !this is protected attribute
    
    def get_name(self):
        return self._name
    def get_age(self):
        return self._age
```
