## Generators in Python

### Generator are simple way to create iterators .They use the yield keyword to produce a series of values lazily which means they generate values on the fly  and do not store them in memory.

```python
def square(n):
    for i in range(n):
        return i**2
```

```python
square(3)
```

```python
def square_generator(n):
    for i in range(n):
        yield i**2
```

```python
square_generator(3)
```

```python
for i in square_generator(3):
    print(i)
```

```python
def local_gen():
    print("first")
    yield 1
    print("second")
    yield 2
    print("Third")
    yield 3
```

```python
gen = local_gen()
```

```python
print(next(gen))
print(next(gen))
print(next(gen))
```

### Reading large file 
#### generators are useful for reading the large file they load onle line at a time without loading entire file

```python
def read_large_file(file_path):
    with open (file_path , 'r') as f:
        for line in f:
            yield line
```

```python
file_path = 'large_File.txt'

for line in read_large_file(file_path=file_path):
    print(line.strip())
```
