```python
'''
__init__: Initializes a new instance of a class.
__str__: Returns a string representation of an object.
__repr__: Returns an official string representation of an object.
__len__: Returns the length of an object.
__getitem__: Gets an item from a container.
__setitem__: Sets an item in a container.
'''
```

```python
class Student:
    def __init__(self, name):
        self.name = name

s = Student("Priyanshu")
print(s.name)
```

```python
class Student:
    def __init__(self, name):
        self.name = name


s = Student("Priyanshu")
print(s)
```

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student Name: {self.name}"

s = Student("Priyanshu")
print(s)
```

```python
class Student:
    def __init__(self, name):
        self.name = name

    

s = Student("Priyanshu")

print(repr(s))
```

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Student('{self.name}')"

s = Student("Priyanshu")

print(repr(s))
```

```python
class Team:
    def __init__(self):
        self.members = ["A", "B", "C"]
# ! you have to implement __len__ method to use len() function on Team class
    def __len__(self):
        return len(self.members)

t = Team()

print(len(t))
```

```python
class Numbers:
    def __init__(self):
        self.data = [10, 20, 30]
# ! you have to implement __getitem__ method to use indexing on Numbers class
    def __getitem__(self, index):
        return self.data[index]

n = Numbers()

print(n[1])
```

```python
class Numbers:
    def __init__(self):
        self.data = [10, 20, 30]

# ! you have to implement __setitem__ method to assign values using indexing on Numbers class

    def __setitem__(self, index, value):
        self.data[index] = value

n = Numbers()

n[1] = 100

print(n.data)
```
