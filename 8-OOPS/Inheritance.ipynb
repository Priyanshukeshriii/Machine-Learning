Inheritance

```python
class Car:
    def __init__(self, windows , doors , enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def get_description(self):
        return f"{self.year} {self.make} {self.model}"
    def drive(self):
        print(f"The person will drive the {self.enginetype} car")


```

```python
Car1 = Car(4, 4, "V6")
print(Car1.windows)
print(Car1.doors)
print(Car1.enginetype)
Car1.drive()
```

inherite the class car

```python
class tesla(Car):
    def __init__(self, windows , doors , enginetype, autopilot):
        super().__init__(windows , doors , enginetype)
        self.autopilot = autopilot

    def drive(self):
        print(f"The person will drive the {self.enginetype} car with autopilot {self.autopilot}")
```

```python
tesla1 = tesla(4, 4, "Electric", "Full Self-Driving")
print(tesla1.windows)
print(tesla1.doors)
print(tesla1.enginetype)
print(tesla1.autopilot)
tesla1.drive()
```

Multi Inheritance

```python
# base class 1
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def make_sound(self):
        print("The animal makes a sound")

# base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner

    def play(self):
        print("The pet is playing")
# derived class
class Dog(Animal, Pet):
    def __init__(self, name, species, owner, breed):
        Animal.__init__(self, name, species)
        Pet.__init__(self, owner)
        self.breed = breed

    def make_sound(self):
        print("The dog barks")
```

```python
# creating an instance of the Dog class
dog1 = Dog("Buddy", "Canine", "Alice", "Golden Retriever")
print(dog1.owner)
dog1.make_sound()
dog1.play()
```
